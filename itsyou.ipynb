```python
import time

# List lirik dengan waktu munculnya (dalam detik)
lirik = [
    (1, "Then, please, don't let history repeat itself"),
    (6, "'Cause I want you yeah, I want you yeah"),
    (12, "There's nothing else I want"),
]

# Catat waktu mulai
start_time = time.time()

# Loop untuk menampilkan lirik sesuai waktu yang diatur
for waktu, teks in lirik:
    while time.time() - start_time < waktu:
        time.sleep(0.1)  # Menunggu sampai waktunya
    print(teks)  # Menampilkan lirik

print("</3")
```
